### What is Python?
Python is a high-level programming language. A high-level language is designed to be read and written by people more easily than lower-level machine instructions. This makes Python a good language for beginners, but it is also powerful enough to be used in professional research and software development.

Python code often looks close to plain English. This is one of the reasons it is widely used in education and research. A readable language makes it easier to understand what a program is doing, return to old code, and share scripts with colleagues.

For example, the following line tells Python to display a message:

````python
print("Hello, world!")
````

Even if you have never programmed before, you can probably make a reasonable guess about what this line does. It prints, or displays, the message `Hello, world!`.

#### Why researchers use Python
Researchers often need to solve problems quickly and accurately. They may need to analyse data, automate calculations, generate plots, or process many files. Python is useful because it allows researchers to write programs quickly while still providing access to powerful scientific tools.

Python can help researchers to:
* Automate repetitive tasks.
* Analyse experimental or simulation data.
* Create figures and visualisations.
* Develop reproducible computational workflows.
* Connect different tools and stages of a research process.

#### Python in research computing
Python is often used alongside specialist scientific libraries. These libraries provide tested, efficient tools for common research tasks. For example, researchers may use libraries for numerical calculations, data analysis, plotting or machine learning.

You do not need to know these libraries yet. In this lesson, the important point is that Python provides a foundation on which more advanced research computing skills can be built.

| Python role | Example use |
| :--- | :--- |
| **Automation** | Rename files, generate job scripts, or repeat calculations. |
| **Data processing** | Read output files and calculate summaries. |
| **Visualisation** | Create plots and figures from results. |
| **Workflow control** | Connect simulation, analysis and reporting steps. |

### Running Python
There are several ways to use Python. In this lesson, we will focus on two common approaches:
* Using Python interactively.
* Writing and running Python scripts.

Both approaches are useful. The best choice depends on what you are trying to do.

#### Which Python to Use
On most HPC systems there will be a default python on your system. This is fine for now, but as you get more acquainted with Python you may want to load a specific version that has all the required libraries you need. It is also good practice when submitting a job to a compute node to specify the python version as well as any other required software. This is done with the standard `module load` command.

#### Interactive Python
Interactive Python allows you to type one command at a time and immediately see the result. This is useful when you are learning, testing a small idea, or checking a quick calculation.

An interactive Python session usually shows a prompt that looks like `>>>`.
You type your command right after the prompt, like the active cell shown below:

````python
print("Hello, world!")
````

The block above is the command you typed. The output produced by Python displays directly beneath it when executed.

#### When to use interactive Python
Interactive Python is useful when you want to:
* Try out a single command.
* Test a small calculation.
* Explore how a function works.
* Learn a new Python idea without creating a file.

However, interactive commands are not normally saved as a reusable program. For larger tasks, or for work that needs to be repeated, it is better to write a script.

#### Python scripts
A Python script is a text file containing Python code. The file usually has a name ending in `.py`. For example, let's create a file named `hello.py` using a quick terminal write command:

*(Execute the active cell below to automatically build the script file inside your local workspace)*

````python
%%writefile hello.py
print("Hello, world!")
````

A script allows you to save your commands, edit them later, and run them again. This is very important in research computing because reproducibility matters. If you save your analysis as a script, you or another researcher can run the same steps again in the future.

You can run the script from the command line using the cell below:

````python
!python hello.py
````

When the script runs, Python reads the file and executes the commands inside it.

#### Interactive Python and scripts compared

| Approach | Best used for | Limitation |
| :--- | :--- | :--- |
| **Interactive Python** | Testing ideas and trying small commands. | Commands are not usually saved as a reusable program. |
| **Python script** | Saving, sharing and repeating a program. | You need to create and manage a file. |
